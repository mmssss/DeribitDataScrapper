```python
import pandas as pd
from dotenv import load_dotenv
from examples.BestAskAndBid.downloadBest import read_data_from_mysql

load_dotenv()
```

```python
def get_timestamp(table_name, change_id, do_print=True):
    df = read_data_from_mysql(
        f'SELECT TIMESTAMP_VALUE FROM {table_name} WHERE CHANGE_ID = {change_id}')
    ts = pd.to_datetime(df.iloc[0].item(), unit='ms')
    if do_print:
        print(f'CHANGE_ID: {change_id}, timestamp: {ts}')
    return ts

def print_max_change_id(table_name):
    df = read_data_from_mysql(f'SELECT MAX(CHANGE_ID) from {table_name}')
    max_change_id = df.iloc[0].item()
    print(f'Max CHANGE_ID: {max_change_id}', end=', ')
    max_ts = get_timestamp(table_name, max_change_id, do_print=False)
    print(f'TIMESTAMP: {max_ts}')
```

```python
books_table_name = 'TABLE_DEPTH_10'
print_max_change_id(books_table_name)
get_timestamp(books_table_name, 1_114_500_000);
```

```python
trades_table_name = 'Trades_table_test'
print_max_change_id(trades_table_name)
get_timestamp(trades_table_name, 69_000_000);
```
