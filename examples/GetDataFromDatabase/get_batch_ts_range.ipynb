```python
import os
import pandas as pd

pd.set_option('display.max_rows', 1000)
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 1000)
```

```python
def get_ts_range(dir_path: str, ts_column: str, ts_unit):
    files = sorted(os.listdir(dir_path))
    file1 = files[0]
    file2 = files[-1]
    table1 = pd.read_hdf(os.path.join(dir_path, file1))
    table2 = pd.read_hdf(os.path.join(dir_path, file2))
    ts_min = pd.to_datetime(table1[ts_column].min(), unit=ts_unit)
    ts_max = pd.to_datetime(table2[ts_column].max(), unit=ts_unit)
    print(f'ts_min: {ts_min} ({file1})')
    print(f'ts_max: {ts_max} ({file2})')
```

```python
BATCH_DIR = '../DATA/books_processed'
```

```python
get_ts_range(BATCH_DIR, ts_column='timestamp', ts_unit='ms')
```
